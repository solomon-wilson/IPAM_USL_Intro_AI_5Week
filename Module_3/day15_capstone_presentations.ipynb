## Why this matters

Building the tool is half the job; the other half is convincing decision-makers it is worth using — honestly, including its limits. This is exactly what you will do when you propose a real tool inside SLPTA or any organisation.

## Concept — your 8-minute story

A good demo follows a simple arc: **the problem → the data → a live demo → the limitations → what's next.** Spend most of your time on the live demo and on being honest about limits — panels reward that.

## Presentation structure (8 minutes)

| Time | Section | What to say |
|------|---------|-------------|
| 1 min | **Problem** | The SLPTA need (tie back to Route R12 or your chosen scenario) |
| 1 min | **Data** | What you used and how you cleaned it |
| 1 min | **Approach** | Classification / regression / clustering / LLM — and why |
| 3 min | **Live demo** | Run it; show one good case and one tricky case |
| 1 min | **Limitations** | Where it is weak; what a human must still check |
| 1 min | **Next steps** | What you'd build with more time / data |

Then **4 minutes Q&A**.

Your capstone is assessed on four dimensions. **Pass threshold: Proficient (2) or above on every dimension.**

| Dimension | Emerging (1) | Proficient (2) | Strong (3) | Exemplary (4) |
